```python
import pandas as pd

df = pd.read_csv("House_Cleaned_data.csv")
df.head()
```

```python
# Define x(features) and y(target/label data)

x = pd.DataFrame (df["total_sqft"])
y = df["price"]
```

```python
# 2.1 split the data into training testing

from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=.20,random_state=45)
```

```python
print(x_train.shape)
print(y_test.shape)
```

```python
# 3 . Load the algorithms and create the object of algorithms
from sklearn.linear_model import LinearRegression

lr = LinearRegression()
```

```python
# 4. Train the model --> find the pattern value from data

lr.fit(x_train,y_train)
```

```python
# coefficient , intercept

print(lr.coef_)
print(lr.intercept_)
```

```python
# Check the training accuracy
lr.score(x_train,y_train)
```

```python
# Check the testing accuracy
lr.score(x_test,y_test)
```

```python
# 5. test then model/predict

y_pred = lr.predict(x_test)
y_pred
```

```python
x = df[["total_sqft","bath","size_BHK","balcony"]]
```

```python
# 2.1 split the data into training testing
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=.20,random_state=45)

# 3. load the algorithm and create the object of algorithms

from sklearn.linear_model import LinearRegression
lr = LinearRegression()

# 4. Train the model --> find the pattern value from data 
lr.fit(x_train,y_train)
```

```python
# coefficient , intercept

print(lr.intercept_)
print(lr.coef_)

# check the testing accuracy
lr.score(x_test,y_test)
```

```python
# load the model 


model =lr
Total_sqrt = int(input("Enter total_sqft: "))
bath = int(input("Enter number of bathrooms: "))
size_bhk = int(input("Enter number of bathrooms (BHK): "))
balcony = int(input("Enter number of balcony: "))

model.predict([[Total_sqrt,bath,size_bhk,balcony]])
```
