```python
# check the distributions
# histogram only for numerical columns -->specialy from continous data

for i in df.columns:
  if df[i].dtype != 'str':
    lab = i+" Skewness: "+str(df[i].skew())
    # print(lab)
    # print(df[i].skew())
    sns.histplot(df[i],kde=True)

    plt.legend([lab])
    plt.show()
```

# Encoding

* Types

1. Label Encoding : Assigns a unique number to each category.

2. Ordinal Encoding : Assigns numbers according to the order/ranking of categories.

```python
from sklearn.preprocessing import LabelEncoder,OrdinalEncoder,OneHotEncoder

label = LabelEncoder()

l = label.fit_transform(df["area_type"])
print(l[:50])
```

```python
label.classes_    # it show the order of label encoder default values
```

```python
## ordinal for apply the order in distinct values

ord = OrdinalEncoder(categories=[['Super built-up  Area','Built-up  Area','Plot  Area','Carpet  Area']])

o = ord.fit_transform(df[["area_type"]])
print(o[:50])

# o = ord.fit_transform(df["area_type"].values.reshape(-1,1))
```

```python
ord.categories_   # it show the order of label encoder default values
```

```python
## show the data with actual data

for i in zip(df["area_type"],o):
    print(i)
```

```python
## Binary number ---> 0,1 if value then 1 and rest of then equal 0

ohe = OneHotEncoder()

o = ohe.fit_transform(df[["area_type"]])
o
```

```python
o.toarray()    # show the data into catual form
```

```python
ohe.get_feature_names_out()   # return the name of OHE columns in sequence
```

```python
area_ohe = pd.DataFrame(o.toarray(),columns=ohe.get_feature_names_out())   # convert into dataframe
area_ohe
```

```python
df[['area_type_Built-up  Area', 'area_type_Carpet  Area',
       'area_type_Plot  Area', 'area_type_Super built-up  Area']] = area_ohe
```
